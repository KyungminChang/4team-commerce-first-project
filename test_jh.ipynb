```python
import pandas as pd
import numpy as np

# 데이터 불러오기
customers = pd.read_csv("data/customer_hm.csv")
transactions = pd.read_csv("data/transactions_hm.csv")
```

```python
# customers 카피본 생성 -> fashion news frequency 버림 -> 변수 customers_copy1 생성
customers_copy1 = customers.drop(columns=["fashion_news_frequency"])
customers_copy1
```

```python
import matplotlib.pyplot as plt

# customers_copy1의 복사본 customers_copy2 생성
customers_copy2 = customers_copy1.copy()
# age의 나이대 파생컬럼을 만들기 전 이상치 확인용 히스토그램
plt.figure(figsize=(6, 6))
plt.hist(customers_copy2["age"], bins=10)
plt.xlabel("age")
plt.ylabel("count")
plt.title("age histogram")
plt.show()
```

```python
# age의 나이대 파생컬럼 "age_group" 생성
customers_copy2["age_group"] = "10대"
customers_copy2.loc[customers_copy2["age"] < 100, "age_group"] = "70대+"
customers_copy2.loc[customers_copy2["age"] <= 69, "age_group"] = "60대"
customers_copy2.loc[customers_copy2["age"] <= 59, "age_group"] = "50대"
customers_copy2.loc[customers_copy2["age"] <= 49, "age_group"] = "40대"
customers_copy2.loc[customers_copy2["age"] <= 39, "age_group"] = "30대"
customers_copy2.loc[customers_copy2["age"] <= 29, "age_group"] = "20대"

customers_copy2["age_group"].value_counts()
```

```python
# transactions 전처리를 위한 카피본 생성
transactions_copy1 = transactions.copy()
```

```python
# t_dat 컬럼 전처리 전 형식 일치 확인
transactions_copy1[transactions_copy1["t_dat"].str.contains("-") == False]
```

```python
# t_dat의 데이터형 문자열 -> 날짜데이터 변환
transactions_copy1["t_dat"] = pd.to_datetime(transactions_copy1["t_dat"])
transactions_copy1["t_dat"]
```

```python
# t_dat의 데이터 '연-월' 형식 변환
transactions_copy1["t_dat"] = transactions_copy1["t_dat"].dt.to_period("M")
```

```python
# 전처리 결과 확인
transactions_copy1["t_dat"].value_counts()
```

```python
# 1. product_code가 중복된 행들을 찾기 (keep=False를 써야 중복된 모든 행이 나옴)
is_duplicated_code = articles_copy.duplicated(subset=["product_code"], keep=False)

# 2. product_group_name이 Unknown인 행 찾기
is_unknown_group = articles_copy["product_group_name"] == "Unknown"

# 3. 두 조건을 모두 만족(&)하는 데이터 추출
target_df = articles_copy[is_duplicated_code & is_unknown_group]

# 4. 결과 확인
print(f"조건에 맞는 데이터 개수: {len(target_df)}개")
print(target_df[["article_id", "product_code", "prod_name", "product_group_name"]])
```

```python
# 2 테이블 조인 후 널값확인
tr_cus = transactions_copy1.merge(customers_copy2, on="customer_id", how="left")
tr_cus.isna().sum()
```

```python
# 3테이블 조인 후 널값확인
tr_cus_ar = tr_cus.merge(articles_test, on="article_id", how="left")
tr_cus_ar
```

```python
# 월별 상품군별 매출 groupby -> 제품군별 날짜에 따른 매출 pivot table 제작

monthly_product_paid = tr_cus_ar.groupby(["t_dat", "product_group_name"])["price"].sum()
monthly_product_paid = monthly_product_paid.to_frame()
pv_mpp = pd.pivot_table(
    monthly_product_paid, index="product_group_name", columns="t_dat", values="price", aggfunc="sum"
)
pv_mpp
```

```python
# 3. 월별 TOP 10 추출
monthly_product_paid = tr_cus_ar.groupby(["t_dat", "product_group_name"], as_index=False)["price"].sum()
monthly_product_paid["rank"] = monthly_product_paid.groupby("t_dat")["price"].rank(ascending=False, method="first")
mpp_top_10 = monthly_product_paid[monthly_product_paid["rank"] <= 10]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.barplot(data=mpp_top_10[mpp_top_10["t_dat"] == "2019-07"], x="price", y="product_group_name")
plt.title("Top 10 (2019-07)")
plt.show()
```

```python
age_product_sale = tr_cus_ar.groupby(["age_group", "product_group_name"])["price"].sum().reset_index(drop=False)
age_product_sale["price"] = age_product_sale["price"].astype(int)

product_map = [
    "Unknown",
    "Bags",
    "Items",
    "Cosmetic",
    "Furniture",
    "Underwear/nightwear",
    "Garment and Shoe care",
    "Interior textile",
]
age_product_sale.loc[age_product_sale["product_group_name"].isin(product_map), "product_group_name"] = "Others"

pv_aps = pd.pivot_table(
    age_product_sale, index="age_group", columns="product_group_name", values="price", aggfunc="sum"
)
pv_aps
```

```python
# 나이대별 Top 10 제품군(rank함수사용)
age_product_sale = age_product_sale.groupby(["age_group", "product_group_name"])["price"].sum().reset_index()

age_product_sale["rank"] = age_product_sale.groupby("age_group")["price"].rank(ascending=False, method="first")

top10_age_product = age_product_sale[age_product_sale["rank"] <= 10]
```

```python
# 나이대별 Top 10 제품군 비중 시각화
import matplotlib.pyplot as plt

plt.rc("font", family="Malgun Gothic")

pv_aps.plot(kind="bar", stacked=True, figsize=(6, 10))
plt.title("Top 10 Products by Age Group", fontsize=16)
plt.xlabel("Age Group")
plt.ylabel("Total")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```

```python
import seaborn as sns

sns.catplot(
    data=top10_age_product,
    x="price",
    y="product_group_name",
    col="age_group",
    kind="bar",
    palette="rainbow",
    hue="product_group_name",
)
```
