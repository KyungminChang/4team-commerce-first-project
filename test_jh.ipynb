```python
# 1. 고객별로 방문한 날짜(t_dat)가 몇 개인지 세기 (nunique()로 중복 날짜 제거)
# customer_id별 t_dat의 고유 개수(nunique) 구하기
un_c_counts = tr_cus_ar.groupby("customer_id")["t_dat"].nunique()

# 2. 재구매 고객(2번 이상 방문)과 1회성 고객 구분
one_more = un_c_counts[un_c_counts > 1].count()  # 재구매 고객 수
one_time = un_c_counts[un_c_counts == 1].count()  # 한번만 구매한 고객 수
total_customers = len(un_c_counts)

# 3. 비율 계산
re_purchase_rate = (one_more / total_customers) * 100

print(f"전체 고객 수: {total_customers:,}명")
print(f"재구매 고객 수 (2회 이상 방문): {one_more:,}명")
print(f"재구매율: {re_purchase_rate:.2f}%")

# 4. 파이 차트로 시각화
plt.rc("font", family="Malgun Gothic")
plt.figure(figsize=(6, 6))
plt.pie([one_more, one_time], labels=["재구매 고객", "1회성 고객"], autopct="%1.1f%%", startangle=90)
plt.title
plt.show()
```

```python
tr_cus_ar_original = tr_cus_ar.copy()
```

```python
color_price2 = tr_cus_ar_original.groupby("perceived_colour_master_name")["price"].sum().nlargest(10).index
top10_color2 = tr_cus_ar_original[tr_cus_ar_original["perceived_colour_master_name"].isin(color_price2)]

plt_color_age2 = top10_color2.groupby(["age_group", "perceived_colour_master_name"])["price"].sum().reset_index()

plt_color_month2 = top10_color2.groupby(["t_dat", "perceived_colour_master_name"])["price"].sum().reset_index()
# (문자열로 변환 안하니까 오류나옴)
```

```python
# 상위 10개 색상
top10_colors3 = tr_cus_ar_original.groupby("perceived_colour_master_name")["price"].sum().nlargest(10).index.tolist()

# 상위 10개 색상이고 악세사리 이면
top_with_acc = tr_cus_ar_original[
    (tr_cus_ar_original["perceived_colour_master_name"].isin(top10_colors3))
    & (tr_cus_ar_original["product_group_name"] == "Accessories")
].copy()

plt_top_acc = top_with_acc.groupby(["t_dat", "perceived_colour_master_name"])["price"].sum().reset_index()
plt_top_acc["t_dat"] = plt_top_acc["t_dat"].astype(str)

plt.figure(figsize=(14, 7))
sns.lineplot(
    data=plt_top_acc, x="t_dat", y="price", hue="perceived_colour_master_name", hue_order=top10_colors3, marker="o"
)

plt.title("Monthly Sales Trend: Top 7 Colors Accessories", fontsize=16)
plt.xticks(rotation=45)
plt.grid(True, axis="y", linestyle="--", alpha=0.5)
plt.legend(title="Top 7 Colors", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```

```python
top10_colors3_df = tr_cus_ar_original[tr_cus_ar_original["perceived_colour_master_name"].isin(color_price2)]
# product_group_name별로 세분화하기 위해 groupby에 추가
plt_div_data = (
    top10_colors3_df.groupby(["t_dat", "product_group_name", "perceived_colour_master_name"])["price"]
    .sum()
    .reset_index()
)
plt_div_data["t_dat"] = plt_div_data["t_dat"].astype(str)

# relplot을 이용해 product_group_name별로 그래프 생성
# col="product_group_name" 필요
prod_group_graph = sns.relplot(
    data=plt_div_data,
    x="t_dat",
    y="price",
    hue="perceived_colour_master_name",
    hue_order=color_price2,
    col="product_group_name",
    kind="line",
    marker="o",
    col_wrap=3,
    height=4,
    aspect=1.2,
    facet_kws={"sharey": False},  # 제품군마다 매출 규모가 다르므로 y축 범위를 각자 조절
)
prod_group_graph.set_titles("{col_name}")  # 그래프 상단에 col="product_group_name"에 해당하는 이름표시
plt.show()
```

```python
# 상위 6개 제품군 리스트
# 매출액(price) 합계 기준 상위 6개 제품군의 이름을 가져옵니다.
top6_product_groups = top10_colors3_df.groupby("product_group_name")["price"].sum().nlargest(6).index.tolist()

# 데이터프레임에서 상위 6개 제품군만
top10_colors3_df = top10_colors3_df[top10_colors3_df["product_group_name"].isin(top6_product_groups)]
plt_div_data = (
    top10_colors3_df.groupby(["t_dat", "product_group_name", "perceived_colour_master_name"])["price"]
    .sum()
    .reset_index()
)
plt_div_data["t_dat"] = plt_div_data["t_dat"].astype(str)

# relplot 그리기
g = sns.relplot(
    data=plt_div_data,
    x="t_dat",
    y="price",
    hue="perceived_colour_master_name",
    hue_order=color_price2,
    col="product_group_name",
    col_order=top6_product_groups,  # 제품군이 나오는 순서도 매출순
    kind="line",
    marker="o",
    col_wrap=3,
    height=4,
    aspect=1.2,
    facet_kws={"sharey": False},
)
g.set_titles("{col_name}", fontweight="bold")

plt.show()
```

실제 데이타 값이 평균을 기준으로 할때 얼마나 들쭉 날쭉하냐를 나타내는 것


```python
# 1. 잘 팔리는 '상위 10개 카테고리' 이름 뽑기
top10_cats = tr_cus_ar.groupby("product_group_name")["price"].sum().nlargest(10).index

# 2. 잘 팔리는 '상위 10개 색상' 이름 뽑기
top10_cols = tr_cus_ar.groupby("perceived_colour_master_name")["price"].sum().nlargest(10).index

# 3. 데이터 필터링 (상위 10개에 해당하는 것만 남기기)
# "카테고리도 Top 10 안에 들고, 색깔도 Top 10 안에 드는 데이터만 가져와!"
df_filtered = tr_cus_ar[
    (tr_cus_ar["product_group_name"].isin(top10_cats)) & (tr_cus_ar["perceived_colour_master_name"].isin(top10_cols))
]

# 4. 요약표(피벗 테이블) 만들기
# 행(index): 카테고리, 열(columns): 색상, 값(values): 매출액 합계
category_color_pivot = pd.pivot_table(
    df_filtered, index="product_group_name", columns="perceived_colour_master_name", values="price", aggfunc="sum"
)

# 결측치(NaN)는 0으로 채우기 (매출이 없다는 뜻이니까)
category_color_pivot = category_color_pivot.fillna(0)

print("표 만들기 완성!")
```

```python
import matplotlib.pyplot as plt

# 1. 그래프 크기 설정
# figsize를 넉넉하게 잡아야 글씨가 안 겹칩니다.
category_color_pivot.plot(kind="bar", stacked=True, figsize=(12, 7), colormap="tab20")

# 2. 제목과 라벨 붙이기
plt.title("카테고리별 색상 매출 비중 (Top 10)", fontsize=15)
plt.xlabel("카테고리 (상품군)")
plt.ylabel("매출액")
plt.xticks(rotation=45)  # 글씨 45도 기울이기
plt.legend(title="색상", bbox_to_anchor=(1.05, 1), loc="upper left")  # 범례를 그래프 밖으로 빼기

# 3. 그래프 보여주기
plt.tight_layout()
plt.show()
```
